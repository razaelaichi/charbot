```python
data = open('words_alpha.txt', 'r')
words = data.read().splitlines()
```

```python
alphabets = set(sorted(list(''.join(words))))
```

```python
i = 0
stoi = {char : i+1 for i,char in enumerate(sorted(alphabets))}
stoi['.'] = 0
```

```python
itos = {i:char for char, i in stoi.items()}
```

```python
import torch
```

```python
base_matrix = torch.zeros(size=(27,27), dtype=torch.int64)
```

```python
for word in words:
    word = "." + word+'.'
    for ch1, ch2 in zip(word, word[1:]):
        x_cord, y_cord = stoi[ch1], stoi[ch2]
        base_matrix[x_cord, y_cord] += 1
```

```python
base_matrix
```

```python
total_per_row = torch.sum(input=base_matrix,dim=1, keepdim= True, dtype=torch.int64)
```

```python
# total_per_row
```

```python
# prob & sum successfully equal to zero
# print(torch.sum(input=(base_matrix / total_per_row), keepdim=True, dim=1))
probs = base_matrix / total_per_row
#probs
```

```python
word = []
current_index = 0
for i in range(100000):
    while True:
        row = probs[current_index]
        sampled_tensor = torch.multinomial(row, num_samples=1).item()
        word += itos[sampled_tensor]
        if sampled_tensor == 0:
            break
        current_index = sampled_tensor
    # limiter that only prints words between 3 and 6 characters long
    if len(word) >= 3 and len(word) <= 6:
        print("".join(word))
    word = []
    current_index = 0
```

```python
x_piles = []
y_piles = []

for word in words:
    word = "." + word + "."
    for ch1, ch2 in zip(word, word[1:]):
        x_piles.append(stoi[ch1])
        y_piles.append(stoi[ch2])

# Convert them into raw PyTorch Tensors
X = torch.tensor(x_piles)
Y = torch.tensor(y_piles)
```

```python
X_enc = torch.nn.functional.one_hot(X, num_classes=27).float()
X_enc.shape
```

```python
W = torch.randn(size=(27,27), dtype=torch.float32, requires_grad=True)
```

```python

logits = X_enc @ W
logits.shape
```

```python
counts = logits.exp()
counts
```

```python
probs_nn = counts / counts.sum(dim = 1, keepdim=True)
probs_nn
```

```python
probs_nn[0].sum()
```

```python
correct_probs = probs_nn[torch.arange(3864812), Y]
loss = (-correct_probs.log().mean())
loss.item()
```

```python
W = torch.randn(size=(27,27), dtype=torch.float32, requires_grad=True)
for i in range(100):
    logits = X_enc @ W
    counts = logits.exp()
    probs_nn = counts / counts.sum(dim = 1, keepdim=True)

    correct_probs = probs_nn[torch.arange(3864812), Y]
    loss = -correct_probs.log().mean()

    loss.backward()

    with torch.no_grad():
        W -= 10 * W.grad
    W.grad.zero_()
    print(f'{loss=}')
```

```python
with torch.no_grad():
    for _ in range(20):
        word = []
        current_index = 0
        while True:
            x = torch.nn.functional.one_hot(torch.tensor([current_index]), num_classes=27).float()
            logits = x @ W
            probs = logits.softmax(dim=1)[0]
            sampled_tensor = torch.multinomial(probs, num_samples=1).item()
            if sampled_tensor == 0:
                break
            word.append(itos[sampled_tensor])
            current_index = sampled_tensor
        if len(word) >= 3:
            print("".join(word))
```
